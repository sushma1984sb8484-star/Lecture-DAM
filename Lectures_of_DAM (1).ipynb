```python
import builtins

# Number Type
# 1. Integer
age = 25
builtins.print(age)
builtins.print(type(age))

# 2. Float
price = 99.50
builtins.print(price)
builtins.print(type(price))

# 3. Complex
z = 3 + 4j
builtins.print(z)
builtins.print(type(z))
```
